```python
import torch
import torch.nn as nn

# Data
X = torch.randn(100, 10)
y = torch.randint(0, 2, (100, 1)).float()

# Model
model = nn.Sequential(
    nn.Linear(10, 5),
    nn.ReLU(),
    nn.Linear(5, 1),
    nn.Sigmoid()
)

loss_fn = nn.BCELoss()
lr = 0.01

# Training loop
for epoch in range(50):
    out = model(X)
    loss = loss_fn(out, y)

    # Manual SGD update
    model.zero_grad()
    loss.backward()

    with torch.no_grad():
        for p in model.parameters():
            p -= lr * p.grad

    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

print("Training completed!")
```
