```python
import torch
import torch.nn as nn
import torch.optim as optim

X = torch.randn(100, 10)
y = torch.randint(0, 2, (100, 1)).float()

for name, Opt in {"SGD": optim.SGD, "Adam": optim.Adam, "RMSprop": optim.RMSprop}.items():

    model = nn.Sequential(
        nn.Linear(10, 5),
        nn.ReLU(),
        nn.Linear(5, 1),
        nn.Sigmoid()
    )

    optimizer = Opt(model.parameters(), lr=0.01)
    loss_fn = nn.BCELoss()

    for _ in range(20):
        loss = loss_fn(model(X), y)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(name, "Loss:", round(loss.item(), 4))
```
