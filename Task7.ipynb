```python
import torch
import torch.nn as nn

class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Conv2d(1, 3, kernel_size=3, padding=1)

    def forward(self, x):
        return self.conv(x)

model = CNN()
x = torch.randn(1, 1, 32, 32)
out = model(x)

print("Output shape:", out.shape)
```
