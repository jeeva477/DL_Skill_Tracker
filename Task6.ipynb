```python
import torch
import torch.nn as nn

class ResNetBlock(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Conv2d(64, 64, 3, padding=1)
        self.relu = nn.ReLU()

    def forward(self, x):
        return self.relu(self.conv(x) + x)

block = ResNetBlock()
x = torch.randn(2, 64, 32, 32)
out = block(x)

print("Input:", x.shape)
print("Output:", out.shape)
```
