```python
import torch
import torch.nn as nn

class RNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.rnn = nn.RNN(10, 20, batch_first=True)
        self.fc = nn.Linear(20, 5)

    def forward(self, x):
        out, _ = self.rnn(x)
        return self.fc(out[:, -1, :])

class LSTM(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(10, 20, batch_first=True)
        self.fc = nn.Linear(20, 5)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1, :])

x = torch.randn(4, 15, 10)

print("RNN Output:", RNN()(x).shape)
print("LSTM Output:", LSTM()(x).shape)
```
