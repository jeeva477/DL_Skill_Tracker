```python
# TASK 1: MLP Forward and Backward Pass

import numpy as np

# Input and target
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

# Weights
W1 = np.random.randn(2, 3) * 0.1
W2 = np.random.randn(3, 1) * 0.1

lr = 0.5

for epoch in range(1000):

    # Forward pass
    h = np.maximum(0, X @ W1)
    out = 1 / (1 + np.exp(-(h @ W2)))

    # Loss
    loss = np.mean((out - y) ** 2)

    # Backward pass
    d2 = 2 * (out - y)
    dW2 = h.T @ d2

    d1 = (d2 @ W2.T) * (h > 0)
    dW1 = X.T @ d1

    # Update weights
    W1 -= lr * dW1
    W2 -= lr * dW2

print("Final predictions:")
print((out > 0.5).astype(int))
```
