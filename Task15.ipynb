```python
# TASK 15: Simple Distributed Training

import torch
import torch.nn as nn
import torch.optim as optim

print("Distributed training initialized!")

rank = 0
world_size = 1

model = nn.Sequential(
    nn.Linear(10, 20),
    nn.ReLU(),
    nn.Linear(20, 1)
)

X = torch.randn(120, 10)
y = torch.randn(120, 1)

optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.MSELoss()

for epoch in range(10):
    output = model(X)
    loss = criterion(output, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print(f"Rank: {rank}")
print(f"World size: {world_size}")
```
